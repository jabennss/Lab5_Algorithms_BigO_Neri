```python
# Generate original incident records
start = ID_SUM
incidents = list(range(start, start + 10))

# Rotate using SEED_DIGIT
rotation = SEED_DIGIT
rotated = incidents[rotation:] + incidents[:rotation]

# Target incident
target = (ID_SUM + ARTIST_SUM) % 100

# Search functions
def linear_search(arr, target):
    comparisons = 0
    for i, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

def rotated_binary_search(arr, target):
    comparisons = 0
    low, high = 0, len(arr) - 1
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, comparisons
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

# Run searches
lin_pos, lin_comp = linear_search(rotated, target)
bin_pos, bin_comp = rotated_binary_search(rotated, target)

# === OUTPUT ===
print("\n=== EXERCISE 1 RESULTS ===")
print(f"Original Incident Records: {incidents}")
print(f"Rotation Point:            {rotation}")
print(f"Rotated Incident Records:  {rotated}")
print(f"Target Incident:           {target}")

print("\n--- Linear Search ---")
print(f"Position:     {lin_pos}")
print(f"Comparisons:  {lin_comp}")
print(f"Result:       {'Found at index ' + str(lin_pos) if lin_pos != -1 else 'Not Found'}")

print("\n--- Rotated Binary Search ---")
print(f"Position:     {bin_pos}")
print(f"Comparisons:  {bin_comp}")
print(f"Result:       {'Found at index ' + str(bin_pos) if bin_pos != -1 else 'Not Found'}")

print("\n=== COMPARISON ===")
print(f"Linear Search Comparisons:  {lin_comp}")
print(f"Binary Search Comparisons:  {bin_comp}")
print(f"Reduction:                   {lin_comp - bin_comp} comparisons saved")

print("\n=== COMPLEXITY ===")
print("Linear Search Time Complexity: O(n)")
print("Binary Search Time Complexity: O(log n)")
print("Space Complexity: O(1) — in-place search")

print("\nFINAL OUTPUT: Target incident", target, 
      "was", ("FOUND" if lin_pos != -1 else "NOT FOUND"),
      ". Binary Search used", bin_comp, "comparisons vs", lin_comp, "for Linear Search.")
```

```python
# Generate datasets
start_a = SEED_DIGIT
size_a = ID_SUM // 3
dataset_a = list(range(start_a, start_a + size_a))

start_b = ARTIST_LENGTH
size_b = NAME_LENGTH * 3
dataset_b = list(range(start_b, start_b + size_b))

# Merge function
def merge_sorted(list_a, list_b):
    i = j = comparisons = 0
    merged = []
    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] < list_b[j]:
            merged.append(list_a[i])
            i += 1
        else:
            merged.append(list_b[j])
            j += 1
    merged.extend(list_a[i:])
    merged.extend(list_b[j:])
    return merged, comparisons

# Run merge
consolidated, total_compares = merge_sorted(dataset_a, dataset_b)
is_sorted = all(consolidated[i] <= consolidated[i+1] for i in range(len(consolidated)-1))

# === OUTPUT ===
print("\n=== EXERCISE 2 RESULTS ===")
print(f"Department A Dataset:  {dataset_a}")
print(f"Department B Dataset:  {dataset_b}")
print(f"Consolidated Dataset: {consolidated}")
print(f"Records Before:       {len(dataset_a)} + {len(dataset_b)} = {len(dataset_a) + len(dataset_b)}")
print(f"Records After:        {len(consolidated)}")
print(f"Total Comparisons:    {total_compares}")
print(f"Order Verification:   {'PASSED ✅' if is_sorted else 'FAILED ❌'} — sorted ascending")

print("\n=== COMPLEXITY ===")
print("Time Complexity:  O(n + m) — linear time")
print("Space Complexity: O(n + m) — merged list")

print("\nFINAL OUTPUT: Successfully consolidated", len(consolidated), 
      "records.", total_compares, "comparisons made. Result is sorted.")
```

```python
# Generate base orders
base_orders = [(ID_SUM + i * 7) % 100 for i in range(8)]

# Create 3 input conditions
sorted_orders = sorted(base_orders)
reverse_orders = sorted(base_orders, reverse=True)
random_orders = base_orders.copy()

# Sorting algorithms
def bubble_sort(data):
    a, comp, move = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comp += 1
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                move += 1
                swapped = True
        if not swapped:
            break
    return a, comp, move

def selection_sort(data):
    a, comp, move = data.copy(), 0, 0
    for i in range(len(a)-1):
        min_idx = i
        for j in range(i+1, len(a)):
            comp += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            move += 1
    return a, comp, move

def insertion_sort(data):
    a, comp, move = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comp += 1
            if a[j] <= key:
                break
            a[j+1] = a[j]
            move += 1
            j -= 1
        a[j+1] = key
    return a, comp, move

# Test helper
def test_sort(name, sort_fn, data):
    start = time.perf_counter()
    result, comp, move = sort_fn(data)
    elapsed = time.perf_counter() - start
    is_correct = result == sorted(data)
    return result, comp, move, elapsed, is_correct

# Run all tests
def run_all_tests(label, data):
    print(f"\n{'='*50}")
    print(f"INPUT CONDITION: {label} → {data}")
    print('='*50)
    for algo_name, fn in [("Bubble Sort", bubble_sort), 
                           ("Selection Sort", selection_sort), 
                           ("Insertion Sort", insertion_sort)]:
        result, comp, move, t, ok = test_sort(algo_name, fn, data)
        status = "✓" if ok else "✗"
        print(f"{algo_name:<15} | Comp:{comp:3d} | Move:{move:3d} | Time:{t:.8f}s | {status}")
    print()

# === OUTPUT ===
print("\n=== EXERCISE 3 RESULTS ===")
print(f"Base Order Dataset:   {base_orders}")
run_all_tests("SORTED", sorted_orders)
run_all_tests("REVERSE", reverse_orders)
run_all_tests("RANDOM", random_orders)

print("=== BIG O ANALYSIS ===")
print("All — Worst Case:   O(n²)")
print("Bubble — Best Case:  O(n)  (adaptive)")
print("Insertion — Best Case: O(n) (adaptive)")
print("Selection — Best Case: O(n²) (not adaptive)")
print("Space Complexity:    O(1) — in-place sorting")

print("\n=== OBSERVATIONS ===")
print("Best Case:  Bubble & Insertion fastest on already-sorted data")
print("Worst Case: All slow — Selection uses fewest moves")
print("Random:     Insertion Sort generally most efficient")

print("\nFINAL OUTPUT: Initial data order strongly affects performance.")
print("Insertion Sort recommended for small/nearly sorted datasets.")
```
